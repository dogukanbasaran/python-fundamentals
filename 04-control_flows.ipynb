###### For Loops

<p>For loops. Used for iterating over a sequence.</p>

```python
courses = [
    "Introduction to Computational Social Sciences",
    "Advanced Data Analysis with Python",
    "Classical Social Theory",
]
```

```python
for course in courses:
    print(course)
```

<p><b>break</b> statement allows to stop the loop in a given position.</p>

```python
for course in courses:
    if "Python" in course: 
        print(course) 
        break
```

<p><b>continue</b> statement allows to stop the current iteration in a given position and continue to iterating with next items.</p>

```python
for course in courses:
    if "Introduction" in course:
        continue
    print(course)
```

<p><b>The range()</b> function generates a sequence of numbers for looping. It starts at 0, increases by 1, and stops before the specified number.</p>

```python
for number in range(10):
    print(number)
```

```python
for number in range(1, 10):
    print(number)
```

```python
for number in range(1, 10, 3):
    print(number)
```

<p>A common use of range is for iterating through sequences by index.</p>

```python
for i in range(len(courses)):
    print(f"{i}: {courses[i]}")
```

###### While Loops

<p>Execute a block of code as long as a condition is True.</p>

```python
number = 1
while number < 10:
    print(number)
    number = number + 1
```

```python
number = 1
while number < 10:
    print(number)
    if number == 5:
        break
    number = number + 1
```

```python
number = 0
while number < 10:
    number = number + 1
    if number % 2 == 0:
        continue
    print(number)
```

##### Conditional Statements

```python
gpa = 3.37
```

```python
if gpa >= 3.50:
    print("High Honor Student")
elif gpa >= 3.25:
    print("Honor Student")
else:
    print("Average Student")
```

```python
another_gpa = 3.68
```

```python
if another_gpa > gpa: print("YES")
```

```python
print("YES") if gpa > another_gpa else print("NO") # called Ternary Operator / Conditional Expression 
```

```python
students_gpa_list = [3.37, 3.68, 3.10, 2.90, 2.10, 3.02, 3.23, 3.78, 2.45, 2.78]
```

```python
high_honor_students_gpas = []
honor_students_gpas = []
average_students_gpas = []
```

```python
for gpa in students_gpa_list:
    if gpa >= 3.50:
        high_honor_students_gpas.append(gpa)
    elif gpa >= 3.25:
        honor_students_gpas.append(gpa)
    else:
        average_students_gpas.append(gpa)
    
```

```python
print(high_honor_students_gpas)
```

```python
print(honor_students_gpas)
```

```python
print(average_students_gpas)
```
