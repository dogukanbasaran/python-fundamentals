## Loops

<p>In Python, there are two loops: for and while.</p>

### For Loop

```python
courses = [
    "Introduction to Computational Social Sciences",
    "Programming for Computational Social Sciences",
    "Data Analysis and Visualization",
    "Computational Methods in Social Sciences",
    "Machine Learning for Social Sciences",
    "Network Analysis",
    "Natural Language Processing for Social Sciences",
    "Classical Social Theory"
]
```

<p>For loop is used for iterating over a sequence. It executes a set of statements for each item in a sequence.</p>

```python
for course in courses:
    print(course)
```

<p><b>break</b> statement allows to stop the loop in a given position.</p>

```python
for course in courses:
    print(course)
    if course.startswith("Machine Learning"):
        break
```

<p><b>continue</b> statement allows to stop the current iteration in a given position and continue to iterating with next items.</p>

```python
for course in courses:
    if course.startswith(("Data", "Machine", "Network", "Natural")):
        continue
    print(course)
```

<p><b>range</b> function allows to loop through a set of code for a specified number of times. It returns a sequence of numbers, starting from 0 by default, and increments by 1 as default, and ends at a specified number.</p>

```python
for number in range(10):
    print(number)
```

```python
for number in range(1, 10):
    print(number)
```

```python
for number in range(1, 10, 3):
    print(number)
```

### While Loop

<p>used in order to execute a set of statements as long as a condition is true.</p>

```python
number = 1
```

```python
while number < 10:
    print(number)
    number = number + 1
```

<p>In while loops, <b>break</b> statement allows to stop loops even if conditions are true.</p>

```python
number = 1
```

```python
while number < 10:
    print(number)
    if number == 5:
        break
    number = number + 1
```

<p><b>continue</b> statement works as in for loops.</p>

```python
number = 0
```

```python
while number < 10:
    number = number + 1
    if number % 2 == 0:
        continue
    print(number)
```
