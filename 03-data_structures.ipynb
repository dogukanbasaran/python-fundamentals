<p>Because tuples are immutable, which means their size and contents cannot change, there are few methods for them.</p>

```python
credits_for_compulsory_courses = (3, 3, 3) # all compulsory courses above have 3 credit.
```

```python
credits_for_compulsory_courses.count(3) # it counts how many 3-credit courses.
```

```python
compulsory_courses.index("SOCI 503: Classical Social Theory") # it finds the first occurence of the specified value.
```

##### Dictionaries

<p>Stores a collection of key-value pairs.</p>

```python
student = {
    "personal_information": ["Doğukan", "Başaran", 26, "Male"],
    "education_information": ["master's degree", "Computational Social Sciences"]
}
```

```python
print(student)
```

```python
type(student)
```

```python
student["personal_information"]
```

```python
student["courses"] = ["CSSM 501: Introduction to Computational Social Sciences", "SOCI 503: Classical Social Theory","CSSM 512: Advanced Data Analysis with Python"]
```

```python
print(student)
```

```python
student.update({
    "skills" : ["Python", "SQL", "Web Scraping", "Data Analysis", "Statistics and Probability", "NumPy", "Pandas", "Matplotlib/Seaborn"]
})
```

```python
print(student)
```

```python
print(student.keys())
```

```python
print(student.values())
```

```python
print(student.items())
```

##### Sets

<p>Unordered collection of unique elements.</p>

```python
set(["Doğukan", "Doğukan", "Başaran", "Başaran"])
```

```python
{1, 1, 1, 2, 2, 3, 3, 3, 3} # set literal
```

```python
{1, 2, 3} == {3, 2, 1}
```

```python
{1, 1, 1, 2, 2, 3, 3, 3, 3} == {3, 3, 3, 3, 3, 2, 1, 1}
```

###### Set Methods

```python
print(personal_info)
```

```python
personal_info.add("Male")
```

```python
print(personal_info)
```

```python
print(personal_info)
```

```python
personal_info.discard(26)
```

```python
personal_info.remove("Male") 
```

```python
print(personal_info)
```

<p>Sets support mathematical set opreations such as union, intersection, difference, etc.</p>

```python
x = {0, 1, 2, 3, 4, 5}
```

```python
y = {3, 4, 5, 6, 7, 8, 9}
```

```python
x.union(y)
```

```python
x | y
```

```python
x.intersection(y)
```

```python
x & y
```

```python
x.difference(y)
```

```python
x - y
```

```python
z = x & y
```

```python
z
```

```python
z.issubset(x)
```

```python
x.issuperset(z)
```

```python
t = {10, 11, 12}
```

```python
t.isdisjoint(x)
```
