#### Comprehensions

##### List Comprehensions

<p>Allow you to concisely create a new list by filtering the elements of a sequence, tranforming the elements passing the filter into one concise expression.</p>

```python
courses = ["CSSM 501: Introduction to Computational Social Sciences", "SOCI 503: Classical Social Theory", "CSSM 502: Advanced Data Analysis with Python"]
```

```python
[c.upper() for c in courses]
```

```python
[c for c in courses if "CSSM" in c]
```

###### Nested List Comprehensions

```python
all_courses = [
    ["CSSM 501: Introduction to Computational Social Sciences", "CSSM 502: Advanced Data Analysis with Python", "SOCI 503: Classical Social Theory"],
    ["CSSM 511: Social Network Analysis", "CSSM 521: Introduction to Machine Learning"],
    ["CSSM 590: Seminar", "CSSM 595: Thesis"]
]
```

```python
[course for courses in all_courses for course in courses if "Introduction" in course]
```

##### Set Comprehensions

```python
{c.upper() for c in courses}
```

##### Dictionary Comprehensions

```python
location_mapping = {index: value for index, value in enumerate(courses)}
```

```python
print(location_mapping)
```
