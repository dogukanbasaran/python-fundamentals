#### Strings

```python
course_name = "CSSM 501: Introduction to Computational Social Sciences"
```

```python
print(course_name)
```

```python
course_description = """
This course provides an applied, non-technical introduction to the methods and ideas of Computational Social Sciences. 
It discusses how new online data sources and the methods used to analyze them can shed new light on old social science questions and asks brand new questions. 
It also examines some of the ethical and privacy challenges of living in a world where big data and algorithmic decision making are becoming more common.
"""
```

```python
print(course_description)
```

```python
"Introduction" in course_name
```

```python
"Programming" not in course_name
```

```python
# string concatenation
print("Course name is " + course_name + "\nCourse description:" + course_description)   
```

```python
# f-string: formatted string literal
course = f"Course name is {course_name} \nCourse description: {course_description}"
```

```python
print(course)
```

##### String Methods

```python
course_2_name = "SOCI 503: Classical Social Theory"
```

```python
course_2_name = course_2_name.replace("SOCI 503", "SOCI 506")
```

```python
course_2_name = course_2_name.replace("Classical", "Contemporary")
```

```python
course_2_name
```

```python
course_2_name = course_2_name.upper()
```

```python
course_2_name
```

```python
course_2_name = course_2_name.lower()
```

```python
course_2_name
```

```python
course_2_name = course_2_name.capitalize()
```

```python
course_2_name
```

<p>Strings are a sequence of unicode characters. Thus, it is possible to slice strings.</p>

```python
course_code = course_2_name[0:8]
```

```python
course_code = course_code.upper()
```

```python
course_code
```

```python
course_2_name = course_2_name.replace("Soci 506", course_code)
```

```python
course_2_name
```

```python
part_of_course_name = course_2_name.split(":")
```

```python
part_of_course_name
```

```python
course_code == part_of_course_name[0]
```

```python
course_title = part_of_course_name[1]
```

```python
course_title = course_title.title()
```

```python
course_2_name
```

```python
course_2_name = course_2_name.replace("contemporary social theory", course_title)
```
